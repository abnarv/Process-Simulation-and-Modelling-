```python
import numpy as np
from scipy.optimize import fsolve
```

```python
F = 10 #m3/hr
V = 20 #m3
k = 0.05 
n = 2
Ca_in = 100 #mol/m3
#Find Ca_out and conversion(X)
# F(Ca_in - Ca_out) = kV(Ca_out)**n  
```

```python
def CSTR(Ca_out, Ca_in, F, V, k, n):
   return F*(Ca_in - Ca_out) - k*V*(Ca_out)**n
initial_guess = 30
sol = fsolve(CSTR, x0 = initial_guess, args = (Ca_in, F, V, k, n))
Ca_out = sol[0]
X = (Ca_in - Ca_out)/Ca_in
print('Concentration at output: ' , Ca_out , 'mol/m3')
print('Conversion: ', X)
```
