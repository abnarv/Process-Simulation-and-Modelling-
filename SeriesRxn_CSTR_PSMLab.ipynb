```python
import numpy as np
from scipy.integrate import solve_ivp
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt
```

```python
Ca_in, Cb_in, Cc_in = 100, 0, 0
F = 10
V = 20
k1 = 0.1
k2 = 0.06
n1, n2 = 1, 1
tau = V/F
```

```python
def get_Ca(Ca_in, k1, tau):
    Ca_out = Ca_in/(k1*tau +1)
    return Ca_out 
def get_Cb(Ca_in, k1, k2, tau):
    Cb_out = k1*tau*Ca_in/((1+k1*tau)*(1+k2*tau))
    return Cb_out 
def get_Cc(Ca_in, k1, k2):
    Cc_out = k1*k2*(tau**2)*Ca_in/((1+k1*tau)*(1+k2*tau))
    return Cc_out 
```

```python
Ca_out = get_Ca(Ca_in,k1, tau)
Cb_out = get_Cb(Ca_in, k1, k2, tau)
Cc_out = get_Cc(Ca_in, k1, k2, tau)
print('Ca_out:' , Ca_out)
print('Cb_out:' , Cb_out)
print('Cc_out:' , Cc_out)
```
